```python
# Célula 3 — escolha a cidade (altere somente estes três valores).
UF = "MG" #@param {type:"string"}
MUNICIPIO = "Varginha" #@param {type:"string"}
LIMITE_POR_CIDADE = 30 #@param {type:"integer"}

_UFS = set("AC AL AP AM BA CE DF ES GO MA MT MS MG PA PB PR PE PI RJ RN RS RO RR SC SP SE TO".split())
UF = UF.strip().upper()
MUNICIPIO = MUNICIPIO.strip()
if UF not in _UFS:
    raise ValueError("UF inválida.")
if not MUNICIPIO or len(MUNICIPIO) > 80 or "," in MUNICIPIO:
    raise ValueError("Use exatamente um município por execução.")
if type(LIMITE_POR_CIDADE) is not int or not 1 <= LIMITE_POR_CIDADE <= 30:
    raise ValueError("Limite máximo de 30 estabelecimentos.")
print("Piloto preparado:", MUNICIPIO, "/", UF, "— até", LIMITE_POR_CIDADE, "registros.")
```

```python
# Célula 4 — confirmar e executar somente na homologação.
from getpass import getpass
from urllib.parse import urlencode
from urllib.request import Request, urlopen
import os
import re
import subprocess
import sys
import unicodedata

_STAGING = "https://prlrxjothrslhswpkoza.supabase.co"
_CONFIRMAR = input("Para importar na HOMOLOGAÇÃO, digite IMPORTAR: ").strip().upper()
if _CONFIRMAR != "IMPORTAR":
    raise RuntimeError("Cancelado: nada foi importado.")

_SECRET = getpass("Chave service_role da HOMOLOGAÇÃO (campo oculto): ").strip()
if not _SECRET:
    raise ValueError("Chave vazia: importação cancelada.")
os.environ["SUPABASE_URL"] = _STAGING
os.environ["SUPABASE_SERVICE_ROLE_KEY"] = _SECRET
del _SECRET
try:
    _cmd = [
        sys.executable, str(_SCRIPT),
        "--states", UF,
        "--cities", MUNICIPIO,
        "--max-records-per-city", str(LIMITE_POR_CIDADE),
        "--max-records-per-state", str(LIMITE_POR_CIDADE),
        "--database-max-mb", "100",
    ]
    subprocess.run(_cmd, check=True, timeout=1800)
    _city_key = "".join(
        x for x in unicodedata.normalize("NFD", MUNICIPIO.casefold())
        if unicodedata.category(x) != "Mn"
    )
    _city_key = " ".join(re.sub(r"[^a-z0-9]+", " ", _city_key).split())
    _query = urlencode({"select":"id", "state":"eq."+UF, "city_key":"eq."+_city_key})
    _request = Request(
        _STAGING + "/rest/v1/overture_places?" + _query,
        headers={
            "apikey": os.environ["SUPABASE_SERVICE_ROLE_KEY"],
            "Authorization": "Bearer " + os.environ["SUPABASE_SERVICE_ROLE_KEY"],
            "Prefer": "count=exact",
        },
        method="HEAD",
    )
    with urlopen(_request, timeout=30) as _response:
        print("Conferência no Supabase STAGING:", _response.headers.get("Content-Range"))
    print("Confira overture_import_progress antes de passar à próxima cidade.")
finally:
    os.environ.pop("SUPABASE_SERVICE_ROLE_KEY", None)
    os.environ.pop("SUPABASE_URL", None)
    print("Chave retirada do ambiente. Desconecte o runtime ao terminar.")
```
